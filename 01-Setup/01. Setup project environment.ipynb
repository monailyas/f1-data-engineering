## Setup the project environment for formula 1

1. Create External Location `databricks-course-ext-dl134-forumla1`
2. Create Catalog `formula1`
3. Create Schemas `landing`, `bronze`, `silver`, `gold`
4. Create volume files in the `landing` schema

```sql
%fs ls 'abfss://formula1@databrickscourseextdl134.dfs.core.windows.net/landing/'
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_course_ext_dl134_formula1
    URL 'abfss://formula1@databrickscourseextdl134.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-course-sc`)
    COMMENT 'external location for the formula1 container';
```

```sql
SHOW CATALOGS
```

### Create main catalog

```sql
CREATE CATALOG  IF NOT EXISTS  formula1
    MANAGED LOCATION 'abfss://formula1@databrickscourseextdl134.dfs.core.windows.net/' 
   COMMENT 'This is the main catalog for formula1 project' ;
```

### Create schemas landing, bronze, silver, and gold 

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@databrickscourseextdl134.dfs.core.windows.net/bronze'; 
CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@databrickscourseextdl134.dfs.core.windows.net/silver'; 
CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@databrickscourseextdl134.dfs.core.windows.net/gold'; 
   
```

accessing schemas

```sql
SELECT current_catalog();
```

```sql
USE CATALOG formula1;
```

###  Create volume files 

```sql
CREATE EXTERNAL VOLUME formula1.landing.files
LOCATION 'abfss://formula1@databrickscourseextdl134.dfs.core.windows.net/landing';
```
