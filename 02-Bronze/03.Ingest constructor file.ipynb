```python
# source and table name 

source_file = f"{landing_folder_path}/constructors.json"
table_name = f"{catalog_name}.{bronze_schema}.constructors"
```

**Step 1 - Read the CSV file using the dataframe reader API**


```python
# define the schema type 

constructors_schema = "constructorId STRING, name STRING,nationality STRING, url STRING"

```

- _or can use the tripple quotes to seperate line by line 
- constructors_schema = """
        constructorID STRING,
        name STRING,
        nationality STRING, 
        url STRING"""_


```python
constructors_df = ( spark.read
            .format('json')
            .option('header', True)
            #.option('inferSchema', 'true')
            .option('mode', 'FAILFAST')
            .schema(constructors_schema)
            .load(source_file)
)



```

```python
display(constructors_df)
```

**Step 2 - Add the Metadata columns**
  - Source file
  - Ingestion Timestamp


```python
constructors_df_final = add_ingestion_metadata(constructors_df)
```

```python
display(constructors_df_final)
```

**Step 3 - Write to bronze delta table**

```python
(
    constructors_df_final
        .write
        .format('delta')
        .mode('overwrite')
        .saveAsTable(table_name)
)
```
